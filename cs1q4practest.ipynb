```python
from IPython.display import clear_output
import time

#Functions
numbers_list = []

def typewriter_print(text):
  """Prints text with a typewriter effect."""
  for char in text:
    print(char, end='', flush=True)
    time.sleep(0.01)
  print()

def solve(num_to_add):
  numbers_list.append(num_to_add)

def analyze_numbers(numbers_list):
  if not numbers_list:
    typewriter_print("No numbers were entered for analysis.")
    return

  # Helper function to check if a number is prime
  def is_prime(n):
    if n < 2:
      return False
    if n == 2:
      return True
    if n % 2 == 0: # All other even numbers are not prime
      return False
    # Check for odd divisors from 3 up to sqrt(n)
    i = 3
    while i * i <= n:
      if n % i == 0:
        return False
      i += 2
    return True

  analysis_results = [] # List to store calculated results for final summary

  while True:
    typewriter_print("\nWhat operation would you like to perform?")
    typewriter_print("Options: average, sum, odd, even, greatest, smallest, asc(ascending), desc(descending), prime(every prime number), Fin")
    operation = input("Enter your choice: ").strip().lower()

    if operation == "fin":
      clear_output()
      typewriter_print("--- Analysis Summary ---")
      for result_str in analysis_results:
        typewriter_print(result_str)
      typewriter_print("Analysis finished!")
      break
    elif operation == "average":
      if numbers_list:
        result = f"Average of numbers: \033[1;3;30m{sum(numbers_list) / len(numbers_list):.2f}\033[0m"
        analysis_results.append(result)
        typewriter_print(result) # Print immediately
      else:
        typewriter_print("Cannot calculate average, no numbers entered.")
    elif operation == "sum":
      result = f"Sum of numbers: \033[1;3;30m{sum(numbers_list)}\033[0m"
      analysis_results.append(result)
      typewriter_print(result) # Print immediately
    elif operation == "odd":
      odd_numbers = [n for n in numbers_list if n % 2 != 0]
      if odd_numbers:
        result = f"Every odd number: \033[1;3;30m{odd_numbers}\033[0m"
        analysis_results.append(result)
        typewriter_print(result) # Print immediately
      else:
        typewriter_print("No odd numbers found.")
    elif operation == "even":
      even_numbers = [n for n in numbers_list if n % 2 == 0]
      if even_numbers:
        result = f"Every even number: \033[1;3;30m{even_numbers}\033[0m"
        analysis_results.append(result)
        typewriter_print(result) # Print immediately
      else:
        typewriter_print("No even numbers found.")
    elif operation == "greatest":
      if numbers_list:
        result = f"Greatest number: \033[1;3;30m{max(numbers_list)}\033[0m"
        analysis_results.append(result)
        typewriter_print(result) # Print immediately
      else:
        typewriter_print("No numbers to find the greatest from.")
    elif operation == "smallest":
      if numbers_list:
        result = f"Smallest number: \033[1;3;30m{min(numbers_list)}\033[0m"
        analysis_results.append(result)
        typewriter_print(result) # Print immediately
      else:
        typewriter_print("No numbers to find the smallest from.")
    elif operation == "asc":
      if numbers_list:
        sorted_asc = sorted(numbers_list)
        result = f"Numbers in ascending order: \033[1;3;30m{sorted_asc}\033[0m"
        analysis_results.append(result)
        typewriter_print(result) # Print immediately
      else:
        typewriter_print("No numbers to sort.")
    elif operation == "desc":
      if numbers_list:
        sorted_desc = sorted(numbers_list, reverse=True)
        result = f"Numbers in descending order: \033[1;3;30m{sorted_desc}\033[0m"
        analysis_results.append(result)
        typewriter_print(result) # Print immediately
      else:
        typewriter_print("No numbers to sort.")
    elif operation == "prime":
      if numbers_list:
        prime_numbers = [n for n in numbers_list if is_prime(n)]
        if prime_numbers:
          result = f"Every prime number: \033[1;3;30m{prime_numbers}\033[0m"
          analysis_results.append(result)
          typewriter_print(result) # Print immediately
        else:
          typewriter_print("No prime numbers found.")
      else:
        typewriter_print("No numbers to check for primality.")
    else:
      typewriter_print("Invalid operation. Please choose from 'average', 'sum', 'odd', 'even', 'greatest', 'smallest', 'asc', 'desc', 'prime', or 'Fin'.")

#Main loop for the entire bot
while True:
  numbers_list = [] # Clear the list for a new session
  clear_output()
  typewriter_print("Hello there! I am number bot, please input a set of numbers please! :3 ")
  print()

  while True:
    try:
      raw_input = input("Number (or 'Y' to finish): ")
      if raw_input.strip().lower() == 'y':
        typewriter_print("Finishing up collecting numbers!")
        break
      elif raw_input.isdigit(): # If it's a digit, process as a number
        num = int(raw_input)
        solve(num)
      else:
        typewriter_print("Invalid input. Please enter a number or 'Y' to finish.")
        continue
    except ValueError:
      typewriter_print("Invalid input. Please enter a number.")
      continue

  clear_output()
  analyze_numbers(numbers_list)

  restart_choice = input("\nDo you want to restart the number bot? (yes/no): ").strip().lower()
  if restart_choice != 'yes':
    typewriter_print("Thank you for using the number bot! Goodbye!")
    break
```
